```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
img_path = '/content/sample_data/lotus.webp'
image = cv2.imread(img_path)
gray_img = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
edges = cv2.Canny(gray_img, 100, 200)
```

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.title('Original Image')
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title('Canny Edge Detection')
plt.imshow(edges, cmap='gray')
plt.axis('off')

plt.show()
```

```python
img = cv2.imread('/content/sample_data/lotus.webp', 0)

img = cv2.GaussianBlur(img, (3, 3), 0)

sobel_x = np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]])
sobel_y = np.array([[-1, -2, -1], [0, 0, 0], [1, 2, 1]])

grad_x = cv2.filter2D(img, cv2.CV_64F, sobel_x)
grad_y = cv2.filter2D(img, cv2.CV_64F, sobel_y)

magnitude = np.sqrt(grad_x**2 + grad_y**2)
magnitude = np.uint8(255 * magnitude / np.max(magnitude))

_, manual_edges = cv2.threshold(magnitude, 50, 255, cv2.THRESH_BINARY)

plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.title('Original Image')
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title('Manual Edge Threshold')
plt.imshow(manual_edges, cmap='gray')
plt.show()
```

```python
sobel_x = cv2.Sobel(gray_img, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(gray_img, cv2.CV_64F, 0, 1, ksize=3)

# Calculate the magnitude of the gradient
sobel_magnitude = np.sqrt(sobel_x**2 + sobel_y**2)

# Normalize and convert to 8-bit for display
sobel_magnitude = cv2.normalize(sobel_magnitude, None, 0, 255, cv2.NORM_MINMAX)
sobel_magnitude = np.uint8(sobel_magnitude)
```

```python
plt.figure(figsize=(8, 6))
plt.title('Sobel Edge Detection')
plt.imshow(sobel_magnitude, cmap='gray')
plt.axis('off')
plt.show()
```
